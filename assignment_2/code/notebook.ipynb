To find the optimal pair of L and *l*, we must focus on maximizing HL (since higher is better for the evaluation) while minimizing elapsed_time (to ensure efficiency). The optimal pair based on the criteria is:
- L=5
- *l*=4
- Average HL = 48.656
- Average Elapsed Time = 2.324 seconds
This pair provides the highest HL value while maintaining the elapsed time reasonable.

From the generated data we can also observe that **higher** L values (e.g., L=5) tend to achieve higher HL values. For example, with L=5 and *l*=4, the highest HL (48.656) is achieved, albeit with a relatively higher computation time (2.324 seconds).
This suggests that increasing L likely leads to a more thorough or complete evaluation, improving HL scores. However, as L increases, the computational cost (elapsed_time) also rises moderately.

On the other hand, **higher** *l* values (e.g., *l*=4) correlate with higher HL values but also result in increased computation time. For example, when *l*=4, the HL is high for both L=5 (48.656) and L=3 (48.471), but the elapsed time exceeds 2 seconds.
This suggests that higher *l* values refine the computation, leading to better scores at the cost of additional time.

As proven by the optimal pair, a **trade-off** exists between HL and the computation time as L and *l* increase:
- Small values (L=3, *l*=1) result in low computation times (0.619 seconds) but lower HL values (42.741).
- Medium values (L=5, *l*=2) achieve a balance, providing a relatively high HL (46.169) with moderate elapsed time (1.232 seconds).
- Larger values (L=5, *l*=4) maximize HL but take the most time.

Regarding trends, while higher L tends to dominate in improving HL, its effectiveness depends on the value of *l*; higher *l* refines the outcome but has diminishing returns if L is small (e.g., L=2).

Summing up, to maximize the evaluation score (HL), the pair L=5 and *l*=4 yields the best results, although it comes with a higher computation time. For scenarios where a balance between evaluation quality and speed is necessary, choosing a slightly lower *l* (e.g., *l*=2) while maintaining L=5 is an efficient strategy.

However, if the focus is optimize time without sacrificing too much on HL, experimenting with L=3 and *l*=1 could offer a good trade-off, providing reasonable evaluation and faster results.

## Comparison between predMinMax and abMinMax

The predMinMax algorithm and the abMinMax algorithm have been compared with all unique combinations of different branch_factor (ranging from 1 to 5) at L=3. It was chosen L=3 because the MLP regressor was trained on a dataset of over 400,000 rows generated with L=3 by making the algorithm play.

```python
read_csv_as_dataframe("./dataset.csv")
```

As previously stated the chosen regressor was a MLP regressor with 3 hidden layers

```python
mlp_regressor, scaler = train_mlp_regressor("./dataset.csv")
```

With this trained model, 500 tests were conducted for each branch factor value. In each test run, both predMinMax and abMinMax were tasked to find the best move in a randomly generated board. Parameters like HL, computation time and H0 were recorded resulting in the following dataframe.

```python
read_csv_as_dataframe("./stats/pred_minmax_vs_ab_minmax.csv")
```

The recorded data were then analyzed and plotted in graphs for each branch_factor.

```python
plot_pred_minmax_vs_ab_minmax_experiment_results_data("./stats/")
```

By analyzing this data we can see that while the computational effort needed for abMinMax is substantial, particularly in scenarios with large decision trees, predMinMax, conversely, is designed for speed and adaptability. By relying on predictive models, it avoids the need for exhaustive evaluation of all possible game states. While this reduction in computation is advantageous, it introduces a trade-off: the algorithm’s accuracy becomes contingent on the reliability and quality of its predictions. Incorrect predictions or poorly designed heuristics can lead to suboptimal decisions, making this approach potentially less robust in guaranteeing the best possible outcome compared to abMinMax.

Focusing on accurancy and optimality, abMinMax guarantees optimal decisions within the constraints of its search depth. By systematically exploring all relevant paths and using alpha-beta pruning to streamline the process, it ensures that no potentially better move is overlooked.
In contrast, predMinMax sacrifices some of this guaranteed accuracy in favor of speed and flexibility. The reliance on predictive heuristics means that the algorithm does not comprehensively evaluate all potential paths, instead focusing only on those deemed promising based on the predictions. While this can lead to significant efficiency gains, it also introduces the risk of suboptimal decisions if the predictions deviate from actual outcomes. This makes predMinMax less suited for deterministic environments where exhaustive evaluation is feasible but highly effective in complex or probabilistic scenarios where exhaustive search is impractical.

In conclusion, while both abMinMax and predMinMax build upon the foundational minimax algorithm, they serve distinct purposes and cater to different types of challenges. abMinMax emphasizes optimality and exhaustive evaluation within feasible computational limits, making it a robust choice for deterministic, rule-based problems. predMinMax prioritizes efficiency and adaptability, leveraging predictions to navigate complex and uncertain decision spaces. The choice between the two depends on the specific requirements of the application, such as the need for guaranteed accuracy versus the necessity for speed and flexibility.

## Conclusions

The data and comparisons presented provide valuable insights into the suitability and performance of the different algorithms for specific scenarios, along with their trade-offs and optimal configurations.

The abMinMax algorithm emerges as a robust choice. Its ability to systematically explore all possible paths, coupled with alpha-beta pruning, ensures optimal decisions within computational limits. This method's efficiency, adaptability, and precision make it the superior option compared to its counterpart, blMinMax. On the other hand, blMinMax demonstrates reliability in simpler problems but lacks the dynamic pruning capabilities of abMinMax, which hampers its performance in more computationally demanding contexts.

The introduction of predictive capabilities, as seen in predMinMax, shifts the focus toward speed and adaptability. By relying on predictive models rather than exhaustive search, predMinMax significantly reduces computation time. However, this comes at the cost of guaranteed accuracy, as its decision-making heavily depends on the quality and reliability of its predictions.

The predictiveblMinMaxl algorithm provides a middle ground between exhaustive search and prediction-based decision-making. While predictiveblMinMaxl leverages predictions to handle larger decision trees efficiently, it does not guarantee optimal decisions. In contrast, blMinMaxl retains its reliability in deterministic problems by using learned evaluations without sacrificing accuracy.

The analysis of the effects of parameters such as L and *l* reveals important trends. Increasing the search depth L generally results in higher heuristic evaluation scores (HL), reflecting a more thorough assessment of the decision tree. However, this improvement comes with a significant increase in computational cost, especially in deeper searches. Similarly, larger values of *l* refine the evaluations, leading to better HL scores. Nonetheless, this improvement diminishes when L is small, indicating that the benefits of higher *l* depend on an adequate search depth.

Branching factor, on the other hand, has a relatively minor impact on computation time for abMinMax and blMinMax, indicating that these algorithms' performance is not heavily influenced by the complexity of the decision tree in this dataset. However, predictive algorithms such as predMinMax and predictiveblMinMaxl are more sensitive to branching, as their reliance on heuristic predictions amplifies the significance of accurate evaluations when the branching factor increases.

The training of regressors introduces another layer of complexity. The comparison between static H0-based regressor and the one trained with H0 function dataset shows that the latter predicts a broader range of heuristic values. However, the computational cost nearly doubles for H0-function-trained models due to the increased complexity of their input data. While static H0-based regressors are faster, the enriched data of H0-function-trained models enhances their potential to handle more nuanced scenarios.

In conclusion, abMinMax and blMinMax performance improves with deeper searches, particularly with L=5, where they leverage pruning and systematic evaluation to ensure robust outcomes. Predictive algorithms, including predMinMax and predictiveblMinMaxl, benefit from carefully tuned parameters such as L and *l*.

Finally, the choice between static and extended regressors depends on the problem's complexity. While static regressors excel in simpler scenarios, extended datasets allow for more comprehensive evaluations, making them invaluable in challenging or dynamic environments. This data underscores the importance of aligning algorithmic choices and parameter tuning with the specific requirements and constraints of the application domain.
